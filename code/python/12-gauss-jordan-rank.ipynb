## In code

MATLAB has `rref` built in. NumPy doesn't, because computers rarely need the reduced form (elimination and back substitution is cheaper), but a short function does it. It chooses the largest available entry as each pivot, which keeps rounding errors small ([Chapter 25](https://trgoodale.github.io/matrix-book/part-5/25-pivoting.html)), and treats tiny numbers as zero.

```python
import numpy as np

def rref(M, tol=1e-10):
    """Reduced row echelon form of M, and the list of pivot columns."""
    R = np.array(M, dtype=float)
    m, n = R.shape
    pivots, r = [], 0
    for c in range(n):
        if r == m:
            break
        p = r + np.argmax(abs(R[r:, c]))   # largest entry at or below row r
        if abs(R[p, c]) < tol:
            continue                       # no pivot in this column
        R[[r, p]] = R[[p, r]]              # swap it into place
        R[r] /= R[r, c]                    # make the pivot 1
        for i in range(m):
            if i != r:
                R[i] -= R[i, c] * R[r]     # clear the rest of the column
        pivots.append(c)
        r += 1
    return R, pivots

Ab = [[1, 2, -1, 3, 4],
      [2, 4, -1, 7, 9],
      [-1, -2, 2, -2, -3]]                 # Example 2
R, piv = rref(Ab)
print(np.round(R, 10) + 0)                 # + 0 turns -0.0 into 0.0
print("pivot columns:", piv)               # counting from 0
```

`np.linalg.matrix_rank` finds the rank directly. It uses a more reliable method than counting pivots (the singular value decomposition, [Chapter 21](https://trgoodale.github.io/matrix-book/part-4/21-svd-pca.html)):

```python
A = np.array(Ab)[:, :4]
print(np.linalg.matrix_rank(A), np.linalg.matrix_rank(Ab))

A2 = np.array([[1, 1, 0], [1, 2, 1], [0, 1, 2]])    # Example 4
print(np.linalg.inv(A2))
```

For exact fractions instead of decimals, the SymPy library (included with Anaconda) has `Matrix(M).rref()`.
