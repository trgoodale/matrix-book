## In code

```python
import numpy as np
import matplotlib.pyplot as plt

def trans(a, b):
    return np.array([[1, 0, a], [0, 1, b], [0, 0, 1.0]])

def rot(deg):
    t = np.radians(deg)
    return np.array([[np.cos(t), -np.sin(t), 0], [np.sin(t), np.cos(t), 0], [0, 0, 1]])

M = trans(2, 1) @ rot(90) @ trans(-2, -1)
print(np.round(M, 10))
print(M @ [3, 1, 1])

# Forward kinematics of Example 4
T = rot(30) @ trans(1.5, 0) @ rot(60) @ trans(1, 0)
print("hand at", np.round(T[:2, 2], 4))

# Apply M to every corner of a house at once: corners as columns
house = np.array([[2, 4, 4, 3, 2, 2], [1, 1, 3, 4, 3, 1], [1, 1, 1, 1, 1, 1]])
moved = M @ house
plt.figure(figsize=(4, 4))
plt.plot(house[0], house[1], color="0.6", lw=2)
plt.plot(moved[0], moved[1], color="#2153b8", lw=2)
plt.plot(2, 1, "ko")
plt.axis("equal"); plt.grid(alpha=0.3)
plt.show()
```
