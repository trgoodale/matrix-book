### Code for Figure 16.1

This draws [Figure 16.1](https://trgoodale.github.io/matrix-book/part-3/16-fundamental-subspaces.html#fig-row-null) in the chapter. You don't need it for the rest of the notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

r1, r2 = np.array([1, 1, 2]), np.array([1, 2, 3])
n = np.array([-1, -1, 1])
fig = plt.figure(figsize=(5.5, 5))
ax = fig.add_subplot(projection="3d")
u1 = r1 / np.linalg.norm(r1)                     # two perpendicular directions in the plane
u2 = np.cross(n, r1); u2 = u2 / np.linalg.norm(u2)
s, t = np.meshgrid(np.linspace(-2.2, 2.2, 2), np.linspace(-2.2, 2.2, 2))
X, Y, Z = (s[..., None] * u1 + t[..., None] * u2).transpose(2, 0, 1)
ax.plot_surface(X, Y, Z, color="#0a7050", alpha=0.25)
for r, lab in ((r1, "row 1"), (r2, "row 2")):
    ax.quiver(0, 0, 0, *r, color="#0a7050", arrow_length_ratio=0.12, lw=2)
    ax.text(*(r * 1.08), lab, color="#0a7050")
ax.plot(*np.outer([-2.2, 2.2], n).T, color="#2153b8", lw=3)
ax.text(*(n * 2.3), "null space", color="#2153b8")
ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
for f in (ax.set_xlim, ax.set_ylim, ax.set_zlim):
    f(-2.5, 2.5)
ax.set_box_aspect((1, 1, 1))                     # equal scales, so right angles look right
ax.view_init(elev=25, azim=-70)
plt.tight_layout()
plt.show()
```

## In code

SciPy's `null_space` and `orth`, and MATLAB's `null` and `orth`, return bases for the null space and column space. They give orthonormal bases (perpendicular vectors of length 1, [Chapter 17](https://trgoodale.github.io/matrix-book/part-3/17-orthogonality-least-squares.html)), computed from the singular value decomposition, so they look different from the bases found by hand. They span the same subspaces, which you can check by testing that $\mathbf{A}$ times each null space vector is zero.

```python
import numpy as np
from scipy.linalg import null_space, orth

A = np.array([[1, 2, 0, 1],
              [2, 4, 1, 4],
              [3, 6, 1, 5]])
r = np.linalg.matrix_rank(A)
N = null_space(A)                  # columns: a basis for N(A)
print("rank", r, " nullity", N.shape[1], " n =", A.shape[1])
print(np.round(A @ N, 12) + 0)     # zero: they really solve Ax = 0
print(orth(A).shape[1])            # dimension of the column space
print(null_space(A.T).ravel())     # left null space: a multiple of (1, 1, -1)
```
