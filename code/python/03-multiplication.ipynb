## In code

This is the most common bug in numerical code, so it is worth stating plainly. In both languages there are two different multiplications, and the one that looks natural is *not* the same in each.

| Operation | Python (NumPy) | MATLAB |
|:--|:--:|:--:|
| Matrix product $\mathbf{A}\mathbf{B}$ | `A @ B` | `A * B` |
| Entry-by-entry product | `A * B` | `A .* B` |
| Power $\mathbf{A}^3$ | `np.linalg.matrix_power(A, 3)` | `A^3` |
| Entry-by-entry power | `A ** 3` | `A .^ 3` |

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])
print(A @ B)                       # matrix product

P = np.array([[3, 1], [2, 4]])
Q = np.array([[1, 2], [0, 1]])
print(P @ Q)
print(Q @ P)                       # different: order matters
print(P * Q)                       # entry by entry: NOT the matrix product
```

The column view of $\mathbf{A}\mathbf{x}$, checked in code:

```python
M = np.array([[2, 1], [1, 3]])
x = np.array([3, 2])
print(M @ x)
print(x[0] * M[:, 0] + x[1] * M[:, 1])   # 3 x column 1 + 2 x column 2
```
