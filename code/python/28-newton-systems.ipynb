## In code

```python
import numpy as np
from scipy.optimize import fsolve

def f(v):
    x, y = v
    return np.array([x**2 + y**2 - 4, x * y - 1])

def J(v):
    x, y = v
    return np.array([[2 * x, 2 * y], [y, x]])

x = np.array([2, 0.5])
for k in range(5):
    print(f"{k}  {x[0]:.12f}  {x[1]:.12f}  {np.linalg.norm(f(x)):.1e}")
    x = x - np.linalg.solve(J(x), f(x))        # solve, don't invert

print(fsolve(f, [2, 0.5]))                     # library solver
```
