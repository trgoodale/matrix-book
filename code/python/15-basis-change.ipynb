## In code

Coordinates come from solving $\mathbf{P}\mathbf{c} = \mathbf{v}$, and the change of basis from solving $\mathbf{P}\mathbf{B} = \mathbf{A}\mathbf{P}$. As usual, it is better to solve than to compute $\mathbf{P}^{-1}$ ([the chapter](https://trgoodale.github.io/matrix-book/part-1/07-inverses.html)).

```python
import numpy as np

P = np.array([[1, 1], [1, -1]])            # basis vectors as columns
v = np.array([5, 1])
print(np.linalg.solve(P, v))               # coordinates of v (Example 2)

A = np.array([[3, 1], [2, 2]])
P = np.array([[1, 1], [1, -2]])
B = np.linalg.solve(P, A @ P)              # P^(-1) A P (Example 4)
print(np.round(B, 10) + 0)
```
