### Code for Figure 9.1

This draws [Figure 9.1](https://trgoodale.github.io/matrix-book/part-1/09-invariant-lines.html#fig-invariant-lines) in the chapter. You don't need it for the rest of the notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[3, 1], [2, 2]])
green, blue, grey = "#0a7050", "#2153b8", "#6b7685"

fig, ax = plt.subplots(figsize=(5.5, 5.5))
t = np.array([-8, 8])
ax.plot(t, t, color=green, lw=6, alpha=0.2, label="y = x: stretched by 4")
ax.plot(t, -2 * t, color=blue, lw=6, alpha=0.2, label="y = −2x: every point fixed")

def move(p, colour):
    q = A @ p
    ax.plot(*p, "o", color=colour, ms=6)
    if np.allclose(p, q):
        return
    ax.annotate("", xy=q, xytext=p,
                arrowprops=dict(arrowstyle="-|>", color=colour, lw=1.6, mutation_scale=14))

for p in ([0.5, 0.5], [1, 1], [-0.5, -0.5], [-1, -1]):
    move(np.array(p), green)
for p in ([1, -2], [-1, 2], [1.5, -3], [-1.5, 3], [0.5, -1]):
    move(np.array(p), blue)
move(np.array([1, 0]), grey)
ax.plot([0, 1], [0, 0], color=grey, lw=1, ls=":")
ax.plot([0, 3], [0, 2], color=grey, lw=1, ls=":")

ax.axhline(0, color="black", lw=0.8); ax.axvline(0, color="black", lw=0.8)
ax.set_xlim(-5, 5); ax.set_ylim(-5, 5); ax.set_aspect("equal")
ax.set_xticks(range(-4, 5, 2)); ax.set_yticks(range(-4, 5, 2))
ax.grid(alpha=0.3); ax.set_xlabel("x"); ax.set_ylabel("y")
ax.legend(loc="lower right", fontsize=9)
plt.tight_layout()
plt.show()
```

### Code for Figure 9.2

This draws [Figure 9.2](https://trgoodale.github.io/matrix-book/part-1/09-invariant-lines.html#fig-cube-rotation) in the chapter. You don't need it for the rest of the notebook.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

# Rotation by 120 degrees about (1,1,1): it cycles the axes x -> y -> z -> x
R = np.array([[0, 0, 1], [1, 0, 0], [0, 1, 0]])
# Shift the cube so the rotation visibly moves it
V = np.array([[x, y, z] for z in (0, 1) for y in (0, 1) for x in (0, 1)], float) + [0.6, 0, 0]
faces = [[0, 1, 3, 2], [4, 5, 7, 6], [0, 1, 5, 4], [2, 3, 7, 6], [0, 2, 6, 4], [1, 3, 7, 5]]

fig = plt.figure(figsize=(5.5, 5))
ax = fig.add_subplot(projection="3d")
for f in faces:
    loop = V[f + [f[0]]]
    ax.plot(*loop.T, ls="--", color="grey", lw=1)
W = V @ R.T
ax.add_collection3d(Poly3DCollection([W[f] for f in faces], facecolor="#0a7050",
                                     alpha=0.25, edgecolor="#0a7050"))
t = np.linspace(-0.3, 2.0, 2)
ax.plot(t, t, t, color="#2153b8", lw=2)
ax.set_xlim(0, 2); ax.set_ylim(0, 2); ax.set_zlim(0, 2)
ax.set_xticks([0, 1, 2]); ax.set_yticks([0, 1, 2]); ax.set_zticks([0, 1, 2])
ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
ax.view_init(elev=22, azim=-50)
ax.set_box_aspect((1, 1, 1))
plt.tight_layout()
plt.show()
```

## In code

To find invariant lines with a computer, ask for the eigenvalues and eigenvectors. Each eigenvector is a direction of an invariant line, and its eigenvalue is the stretch factor. Python's `np.linalg.eig` returns the eigenvalues as an array and the eigenvectors as the columns of a matrix, in the same order.

```python
import numpy as np

A = np.array([[3, 1], [2, 2]])
values, vectors = np.linalg.eig(A)
values, vectors = values.real, vectors.real   # recent NumPy returns complex numbers
for lam, v in zip(values, vectors.T):
    print(f"lambda = {lam:.4g}, direction {np.round(v, 4)}, slope {v[1] / v[0]:.4g}")

# The axis of a rotation: the invariant points, found with an eigenvector
C = np.array([[0, 0, 1], [1, 0, 0], [0, 1, 0]])
values, vectors = np.linalg.eig(C)
k = np.argmin(abs(values - 1))           # the eigenvalue equal to 1
print(np.real(vectors[:, k]))            # a multiple of (1, 1, 1)
```

The directions come out with length 1, so they are multiples of $(1, 1)$ and $(1, -2)$ rather than those vectors themselves. The slopes, 1 and $-2$, match Example 3. Recent versions of NumPy return the results as complex numbers even when they are real, which is why the code keeps only the real parts. The rotation $\mathbf{C}$ also has two complex eigenvalues, which is why the code picks out the one equal to 1.
