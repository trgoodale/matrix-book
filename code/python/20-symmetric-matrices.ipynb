## In code

For symmetric matrices, use `np.linalg.eigh` rather than `eig`. It is faster, guarantees real eigenvalues (in increasing order) and orthonormal eigenvectors, and is not upset by tiny rounding errors that make a matrix very slightly unsymmetric. MATLAB's `eig` detects symmetric matrices itself. Cholesky factorisation fails, with an error, exactly when a matrix is not positive definite, which makes it the standard test. Note that MATLAB's `chol` returns the *upper* triangular factor $\mathbf{R} = \mathbf{L}^{\mathsf{T}}$.

```python
import numpy as np

A = np.array([[5, 2], [2, 2]])
lam, Q = np.linalg.eigh(A)            # eigenvalues in increasing order
print(lam)
print(Q)
print(np.allclose(Q.T @ Q, np.eye(2)), np.allclose(Q @ np.diag(lam) @ Q.T, A))

B = np.array([[4, 2, 2], [2, 5, 3], [2, 3, 6]])
print(np.linalg.cholesky(B))          # L with B = L L^T (Example 3)

try:
    np.linalg.cholesky(np.array([[1, 2], [2, 1]]))
except np.linalg.LinAlgError as err:
    print("Not positive definite:", err)
```
