## In code

`np.linalg.norm(x, p)` and `np.linalg.cond(A, p)` compute norms and condition numbers, with `p` equal to `1`, `2` (the default) or `np.inf`; in MATLAB they are `norm(x, p)` and `cond(A, p)`. Computing $\kappa$ exactly needs $\mathbf{A}^{-1}$ or the SVD, which costs more than solving the system, so solvers estimate it cheaply instead: MATLAB's `rcond` gives an estimate of $1/\kappa_1$, and MATLAB warns you when it is close to $\varepsilon$.

```python
import numpy as np
from scipy.linalg import hilbert

print(0.1 + 0.2 == 0.3, np.finfo(float).eps)

x = 1e-8
print((1 - np.cos(x)) / x**2, 2 * np.sin(x / 2)**2 / x**2)   # cancellation

A = np.array([[1, 1], [1, 1.01]])
print(np.linalg.cond(A, np.inf), np.linalg.cond(A))

for n in [4, 8, 12]:
    H = hilbert(n)
    xtrue = np.ones(n)
    xhat = np.linalg.solve(H, H @ xtrue)
    print(f"n = {n:2d}: cond = {np.linalg.cond(H):.1e}, "
          f"error = {np.linalg.norm(xhat - xtrue, np.inf):.1e}")
```
