## In code

```python
import numpy as np

A = np.array([[4, 7], [2, 6]])
b = np.array([1, 2])

print(np.linalg.inv(A))          # the inverse, if you really need it
print(np.linalg.solve(A, b))     # the right way to solve Ax = b
```

The difference matters for large systems. Here both approaches solve a random $2000 \times 2000$ system, and we compare the time taken and the size of the error, measured by how far $\mathbf{A}\mathbf{x}$ is from $\mathbf{b}$:

```python
import time
rng = np.random.default_rng(2)
n = 2000
M = rng.standard_normal((n, n))
v = rng.standard_normal(n)

t0 = time.perf_counter(); x1 = np.linalg.solve(M, v); t1 = time.perf_counter()
x2 = np.linalg.inv(M) @ v;                             t2 = time.perf_counter()
print(f"solve: {t1 - t0:.3f} s, error {np.linalg.norm(M @ x1 - v):.1e}")
print(f"inv:   {t2 - t1:.3f} s, error {np.linalg.norm(M @ x2 - v):.1e}")
```

Using the inverse typically takes several times as long and gives a larger error.
