## In code

Adding partial pivoting to the elimination function from [the chapter](https://trgoodale.github.io/matrix-book/part-2/11-gaussian-elimination.html) takes three lines: find the row with the largest entry in the pivot column, and swap it into place, in both $\mathbf{A}$ and $\mathbf{b}$.

```python
import numpy as np
import scipy.linalg

def solve_ge(A, b, pivot=True):
    """Solve Ax = b by Gaussian elimination, with or without partial pivoting."""
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)
    for k in range(n - 1):
        if pivot:
            p = k + np.argmax(np.abs(A[k:, k]))   # row of the largest entry
            A[[k, p]] = A[[p, k]]                 # swap rows k and p
            b[[k, p]] = b[[p, k]]
        for i in range(k + 1, n):
            m = A[i, k] / A[k, k]
            A[i, k:] -= m * A[k, k:]
            b[i] -= m * b[k]
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - A[i, i + 1:] @ x[i + 1:]) / A[i, i]
    return x

A = [[1e-20, 1], [1, 1]]
b = [1, 2]
print(solve_ge(A, b, pivot=False))         # wrong: [0, 1]
print(solve_ge(A, b, pivot=True))          # right: [1, 1]

A = np.array([[1, 2, 1], [2, 6, 1], [4, 8, 6]])
P, L, U = scipy.linalg.lu(A)               # note: A = P @ L @ U
print(P.T)                                 # the P of P A = L U
print(L)
print(U)
```
