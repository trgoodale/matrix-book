## In code

```python
import numpy as np

# Example 1: the steady state of the weather
M = np.array([[0.7, 0.3, 0.2], [0.2, 0.4, 0.3], [0.1, 0.3, 0.5]])
values, vectors = np.linalg.eig(M)
values, vectors = values.real, vectors.real   # all real here; eig returns complex arrays
k = np.argmax(values)                          # the eigenvalue 1
steady = vectors[:, k] / vectors[:, k].sum()   # scale to add to 1
print(np.round(values, 4), steady)

# Example 2: PageRank by the power method
L = np.array([[0, 0, 1, 0],   # L[i, j] = 1 if page j links to page i
              [1, 0, 0, 0],
              [1, 1, 0, 1],
              [0, 0, 0, 0]])
n, d = 4, 0.85
Mw = L / L.sum(axis=0)        # divide each column by its number of links
x = np.ones(n) / n
for k in range(100):
    x = d * (Mw @ x) + (1 - d) / n
print(x)

# Example 3: an absorbing chain
Q = np.array([[0.1, 0, 0], [0.8, 0.1, 0], [0, 0.85, 0.1]])
R = np.array([[0, 0, 0.85], [0.1, 0.05, 0.05]])
N = np.linalg.inv(np.eye(3) - Q)   # the fundamental matrix
print(N.sum(axis=0))               # expected years, from each starting year
print(R @ N)                       # P(graduate), P(leave), from each starting year
```
