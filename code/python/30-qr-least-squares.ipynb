## In code

`np.linalg.qr(A)` returns the reduced factorisation by default, and MATLAB's `qr(A, 0)` (or `qr(A, "econ")`) does the same. For least squares, use `np.linalg.lstsq` (which uses the SVD), `scipy.linalg.lstsq`, or MATLAB's backslash `A \ b`, which uses Householder QR for a non-square $\mathbf{A}$.

```python
import numpy as np

t = np.linspace(0, 1, 50)
for deg in [5, 8, 10, 12]:
    A = np.vander(t, deg + 1, increasing=True)    # columns 1, t, t^2, ...
    c = np.ones(deg + 1)
    b = A @ c
    c_normal = np.linalg.solve(A.T @ A, A.T @ b)
    Q, R = np.linalg.qr(A)
    c_qr = np.linalg.solve(R, Q.T @ b)            # R is triangular
    print(f"degree {deg:2d}: cond(A) = {np.linalg.cond(A):.1e}, "
          f"error: normal {np.max(abs(c_normal - c)):.1e}, QR {np.max(abs(c_qr - c)):.1e}")
```
