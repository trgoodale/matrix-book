## In code

`np.linalg.eig` and MATLAB's `eig` return the eigenvalues, and the eigenvectors as the columns of a matrix in the same order, each scaled to length 1. They don't use the characteristic polynomial, which would be slow and inaccurate for large matrices. Instead they use the QR algorithm of [Chapter 29](https://trgoodale.github.io/matrix-book/part-5/29-eigenvalue-algorithms.html). The order of the eigenvalues, and the sign of each eigenvector, may differ from a hand calculation.

```python
import numpy as np

A = np.array([[4, 1], [2, 3]])
values, vectors = np.linalg.eig(A)
values, vectors = values.real, vectors.real   # recent NumPy returns complex numbers
print(values)
print(vectors)                   # columns: (1, 1) and (1, -2), scaled to length 1

print(np.allclose(A @ vectors, vectors * values))   # A v = lambda v for each column

print(np.linalg.eigvals(np.array([[1, -1], [1, 1]])))   # complex: 1 + i and 1 - i
```
