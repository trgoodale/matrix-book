## In code

Jacobi and Gauss–Seidel take only a few lines. For real problems, use the library Krylov solvers: `scipy.sparse.linalg.cg` and `gmres` in Python, `pcg` and `gmres` in MATLAB.

```python
import numpy as np
from scipy.linalg import solve_triangular
from scipy.sparse.linalg import cg

n = 50
A = 2 * np.eye(n) - np.eye(n, k=1) - np.eye(n, k=-1)
b = np.ones(n) / (n + 1)**2
D = np.diag(np.diag(A))
L = np.tril(A, -1)
U = np.triu(A, 1)

rho = max(abs(np.linalg.eigvals(np.linalg.solve(D, L + U))))
w = 2 / (1 + np.sqrt(1 - rho**2))
splittings = {"Jacobi": (D, L + U), "Gauss-Seidel": (D + L, U),
              "SOR": (D / w + L, U + (1 - 1 / w) * D)}

for name, (M, N) in splittings.items():
    x = np.zeros(n)
    k = 0
    while np.linalg.norm(b - A @ x) >= 1e-6 * np.linalg.norm(b):
        x = solve_triangular(M, b - N @ x, lower=True)   # M is diagonal or lower triangular
        k += 1
    print(f"{name:13s} {k:5d} steps")

steps = []
x, info = cg(A, b, rtol=1e-6, callback=lambda xk: steps.append(1))
print(f"{'CG':13s} {len(steps):5d} steps")
```
