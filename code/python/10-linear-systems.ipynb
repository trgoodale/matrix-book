### Code for a figure

This draws a figure in the chapter. You don't need it for the rest of the notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

blue, orange, green = "#2153b8", "#9c5208", "#0a7050"
a1, a2, b = np.array([3, 4]), np.array([2, -5]), np.array([7, -6])
x, y = 1, 2

fig, (L, R) = plt.subplots(1, 2, figsize=(9, 4.6))

# Row picture: one line per equation
t = np.linspace(-2, 4, 2)
# grey lines, so that blue and orange can mean "x" and "y" in the column picture
L.plot(t, (7 - 3 * t) / 2, color="0.25", lw=2.2, label="3x + 2y = 7")
L.plot(t, (4 * t + 6) / 5, color="0.45", lw=2.2, ls="--", label="4x − 5y = −6")
L.plot(x, y, "o", color=green, ms=8)
L.annotate("(1, 2)", (x, y), xytext=(10, -14), textcoords="offset points", color=green, fontweight="bold")
L.set_xlim(-2, 4); L.set_ylim(-2, 5)
L.set_title("Row picture: where do the lines cross?", fontsize=11)
L.legend(loc="upper right", fontsize=9)

# Column picture: one arrow per column, then the path to b
def arrow(ax, p, q, colour, lw=2.2, alpha=1.0):
    ax.annotate("", xy=q, xytext=p, arrowprops=dict(arrowstyle="-|>", color=colour, lw=lw,
                                                     alpha=alpha, mutation_scale=14, shrinkA=0, shrinkB=0))

O = np.zeros(2)
arrow(R, O, a2, orange, alpha=0.45)
R.annotate("a₂ = (2, −5)", a2, xytext=(8, -8), textcoords="offset points", color=orange)
arrow(R, O, x * a1, blue, lw=3)
R.annotate("a₁ = (3, 4)", a1, xytext=(6, 4), textcoords="offset points", color=blue)
arrow(R, x * a1, x * a1 + y * a2, orange, lw=3)
R.annotate("2a₂", x * a1 + 0.5 * y * a2, xytext=(8, 0), textcoords="offset points", color=orange)
R.plot(*b, "o", color=green, ms=8)
R.annotate("b = (7, −6)", b, xytext=(-40, -20), textcoords="offset points", color=green, fontweight="bold")
R.set_xlim(-2, 9); R.set_ylim(-8, 6)
R.set_title("Column picture: 1 a₁ + 2 a₂ = b", fontsize=11)

for ax in (L, R):
    ax.axhline(0, color="black", lw=0.8); ax.axvline(0, color="black", lw=0.8)
    ax.set_aspect("equal"); ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## In code

Both languages solve square systems with one command, which uses the elimination method of [the chapter](https://trgoodale.github.io/matrix-book/part-2/11-gaussian-elimination.html). Always check the answer by multiplying back: the **residual** $\mathbf{A}\mathbf{x} - \mathbf{b}$ should be zero, apart from tiny rounding errors.

```python
import numpy as np

A = np.array([[0.6, 0.8, 0.5],
              [0.3, 0.1, 0.2],
              [0.1, 0.1, 0.3]])
b = np.array([60, 22, 18])

x = np.linalg.solve(A, b)
print(x)
print(A @ x - b)          # the residual: zero, up to rounding
```

When $\det\mathbf{A} = 0$, `np.linalg.solve` stops with an error rather than guess:

```python
A2 = np.array([[1, 1, 1], [1, -1, 1], [2, 0, 2]])
b2 = np.array([3, 1, 5])
try:
    np.linalg.solve(A2, b2)
except np.linalg.LinAlgError as err:
    print("Error:", err)
```

To tell "no solution" from "infinitely many", compare the rank of $\mathbf{A}$ with the rank of the augmented matrix. The rank counts the genuinely independent equations ([the chapter](https://trgoodale.github.io/matrix-book/part-2/12-gauss-jordan-rank.html) explains it fully). If adding $\mathbf{b}$ increases the rank, $\mathbf{b}$ brings in an equation that contradicts the others, and there is no solution:

```python
Ab2 = np.column_stack([A2, b2])          # the augmented matrix [A | b]
print(np.linalg.matrix_rank(A2), np.linalg.matrix_rank(Ab2))
```
