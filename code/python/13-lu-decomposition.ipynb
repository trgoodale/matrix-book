## In code

SciPy, the scientific companion to NumPy, has the factorisation in `scipy.linalg`. Watch out for one difference of convention: SciPy's `lu` returns $\mathbf{P}$, $\mathbf{L}$, $\mathbf{U}$ with $\mathbf{A} = \mathbf{P}\mathbf{L}\mathbf{U}$, so its $\mathbf{P}$ is the transpose of the $\mathbf{P}$ in $\mathbf{P}\mathbf{A} = \mathbf{L}\mathbf{U}$. MATLAB's `lu` follows the convention of this chapter.

```python
import numpy as np
import scipy.linalg as sl

A = np.array([[2, 1, 1], [4, 3, 3], [8, 7, 9]])
P, L, U = sl.lu(A)                 # A = P @ L @ U
print(P)
print(np.round(L, 4))
print(np.round(U, 4))
print(np.allclose(P @ L @ U, A))
```

Partial pivoting chose 8, the largest entry in column 1, as the first pivot, so these factors differ from the ones found by hand. For solving, use `lu_factor` once and `lu_solve` for each right-hand side:

```python
factors = sl.lu_factor(A)          # the expensive step, done once
for b in ([4, 10, 24], [1, 1, 1]):
    print(np.round(sl.lu_solve(factors, b), 10) + 0)
```

The saving is real. Here 100 right-hand sides are solved for a $1000 \times 1000$ matrix, first by solving from scratch each time, then by factorising once:

```python
import time
rng = np.random.default_rng(3)
n, k = 1000, 100
M = rng.standard_normal((n, n))
B = rng.standard_normal((n, k))

t0 = time.perf_counter()
for j in range(k):
    np.linalg.solve(M, B[:, j])
t1 = time.perf_counter()
f = sl.lu_factor(M)
for j in range(k):
    sl.lu_solve(f, B[:, j])
t2 = time.perf_counter()
print(f"solve each time: {t1 - t0:.2f} s,  factorise once: {t2 - t1:.2f} s")
```
