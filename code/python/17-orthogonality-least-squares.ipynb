## In code

For least squares, use a solver that works on $\mathbf{A}$ directly rather than forming $\mathbf{A}^{\mathsf{T}}\mathbf{A}$: NumPy's `lstsq`, or MATLAB's backslash, which automatically gives the least-squares solution when $\mathbf{A}$ has more rows than columns. They use the QR factorisation, which is more accurate ([Chapter 30](https://trgoodale.github.io/matrix-book/part-5/30-qr-least-squares.html)). For fitting polynomials there is also `polyfit`, which lists the coefficients from the highest power down.

```python
import numpy as np

t = np.array([0, 1, 2, 3])
y = np.array([1, 2, 4, 5])
A = np.column_stack([np.ones_like(t), t])        # columns: 1 and t

x, sse, rank, sv = np.linalg.lstsq(A, y, rcond=None)
print(x, sse)                                    # c and m, and the sum of squared errors

print(np.linalg.solve(A.T @ A, A.T @ y))         # the normal equations: same answer
print(np.polyfit(t, y, 1))                       # slope first, then intercept

Q, R = np.linalg.qr(np.array([[1, 1], [1, 0], [0, 1]]))
print(np.round(Q, 4) + 0)                        # orthonormal columns (Example 3, up to sign)
```
