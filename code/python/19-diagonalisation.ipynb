## In code

The diagonalisation comes straight from `eig`. For powers with a specific $k$, `matrix_power` (or `A^k` in MATLAB) multiplies efficiently by repeated squaring; the formula [Equation 19.2](https://trgoodale.github.io/matrix-book/part-4/19-diagonalisation.html#eq-powers) is most useful for understanding.

```python
import numpy as np

A = np.array([[4, 1], [2, 3]])
lam, P = np.linalg.eig(A)
lam, P = lam.real, P.real                       # recent NumPy returns complex numbers
D = np.diag(lam)
print(np.allclose(A, P @ D @ np.linalg.inv(P)))  # A = P D P^(-1)

k = 5
print(np.linalg.matrix_power(A, k))
print(np.round(P @ np.diag(lam**k) @ np.linalg.inv(P)))   # the same, from the eigenvalues

F = np.array([[1, 1], [1, 0]])
print(np.linalg.matrix_power(F, 30)[0, 1])      # the 30th Fibonacci number
```
