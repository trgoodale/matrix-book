## In code

Python's built-in `pow(a, -1, 26)` finds an inverse modulo 26 (and raises an error if there is none). In MATLAB, `[g, u] = gcd(a, 26)` gives the greatest common divisor `g`, and when `g` is 1, `u` is an inverse of `a`.

```python
import numpy as np

def to_nums(text):
    return [ord(ch) - ord("A") for ch in text]

def to_text(nums):
    return "".join(chr(int(n) + ord("A")) for n in nums)

def hill(K, message):
    """Encrypt (or, with the inverse key, decrypt) a message of even length."""
    pairs = np.array(to_nums(message)).reshape(-1, 2).T     # pairs as columns
    return to_text((K @ pairs % 26).T.ravel())

def inverse_mod26(K):
    det = round(np.linalg.det(K)) % 26
    d_inv = pow(det, -1, 26)               # error if gcd(det, 26) != 1
    adj = np.array([[K[1, 1], -K[0, 1]], [-K[1, 0], K[0, 0]]])
    return d_inv * adj % 26

K = np.array([[3, 3], [2, 5]])
c = hill(K, "HELP")
Kinv = inverse_mod26(K)
print(c, hill(Kinv, c))
print(Kinv)

# Known-plaintext attack (Example 3)
P = np.array([[7, 11], [4, 15]])
C = np.array([[7, 0], [8, 19]])
print(C @ inverse_mod26(P) % 26)
```
