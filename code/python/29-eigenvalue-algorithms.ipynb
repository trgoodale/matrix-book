## In code

For large sparse matrices, when only a few eigenvalues are needed, use `scipy.sparse.linalg.eigs` (or `eigsh` for symmetric matrices) in Python and `eigs` in MATLAB. They use **Krylov** methods, Lanczos and Arnoldi, which extract far more from the sequence $\mathbf{q}, \mathbf{A}\mathbf{q}, \mathbf{A}^2\mathbf{q}, \dots$ than the power method does, and with the option `sigma` they use shifted inverse iteration to find the eigenvalues nearest a given value.

```python
import numpy as np
import scipy.linalg

A = np.array([[2.0, 1.0], [1.0, 3.0]])

# Power method
q = np.array([1.0, 0.0])
for k in range(1, 6):
    y = A @ q
    q = y / np.linalg.norm(y)
    print(f"{k}  {q @ A @ q:.10f}")          # Rayleigh quotient

# Shifted inverse iteration: factorise once, solve at each step
lu = scipy.linalg.lu_factor(A - 1.0 * np.eye(2))
q = np.array([1.0, 0.0])
for k in range(1, 5):
    y = scipy.linalg.lu_solve(lu, q)
    q = y / np.linalg.norm(y)
    print(f"{k}  {q @ A @ q:.10f}")

# Basic QR algorithm
Ak = A.copy()
for k in range(5):
    Q, R = np.linalg.qr(Ak)
    Ak = R @ Q
print(np.round(Ak, 4))
```
