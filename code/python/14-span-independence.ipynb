## In code

Both tests come down to rank. To find a dependence relation, find the solutions of $\mathbf{A}\mathbf{c} = \mathbf{0}$: SciPy's `null_space` and MATLAB's `null` do this. They return a vector of length 1, so divide by one of its entries to get tidier numbers.

```python
import numpy as np
from scipy.linalg import null_space

v1, v2, v3 = [1, 0, 1], [2, 1, 0], [0, 1, -2]
A = np.column_stack([v1, v2, v3])          # the vectors as columns
print(np.linalg.matrix_rank(A))            # 2 < 3, so dependent

c = null_space(A)                          # solutions of A c = 0
print(c.ravel() / c[2, 0])                 # scaled so that c3 = 1

b = [1, 2, 3]                              # is b in the span?
print(np.linalg.matrix_rank(A), np.linalg.matrix_rank(np.column_stack([A, b])))
```

The rank is 2, so the three vectors are dependent, with $2\mathbf{v}_1 - \mathbf{v}_2 + \mathbf{v}_3 = \mathbf{0}$ as in Example 2. Adding $\mathbf{b}$ raises the rank to 3, so $\mathbf{b}$ is not in their span.
